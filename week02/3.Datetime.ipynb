## 데이터 형식 처리 (Datetime)

> 데이터를 불러오면 날짜·시간 컬럼이 **문자열(object)** 로 인식되는 경우가 많다.

> 이 상태에서는 연·월·일 추출, 날짜 간 차이 계산, 시계열 정렬 등 **날짜 고유의 연산이 불가능**하므로, 반드시 적절한 데이터 형식으로 변환해야 한다.

#### 왜 형식 변환이 중요한가?
- **연산 가능**: datetime 형식이어야 날짜 간 차이, 비교, 정렬 등이 가능하다.
- **피처 엔지니어링**: 연도·월·요일·분기 등을 추출하여 모델에 유의미한 파생 변수를 생성할 수 있다.
- **메모리 효율**: 문자열보다 datetime 형식이 메모리를 효율적으로 사용하고 처리 속도도 빠르다.

ref: https://blog.naver.com/wideeyed/221603778414

```python
import pandas as pd

df = pd.read_csv('./dataset/datetime.csv')

df.head()
```

```python
df.dtypes
```

```python
## 해당 dataset에서 년도만 추출하려면 어떻게 해야할까?

df['date_1'].str.slice(0,4)
```

#### object -> datetime 형식으로 변경

```python
df['date_1'] = pd.to_datetime(df['date_1'], format='%Y-%m-%d')
df['date_2'] = pd.to_datetime(df['date_2'], format='%Y/%m/%d')
df['date_3'] = pd.to_datetime(df['date_3'], format='%d-%b-%Y')
```

```python
df.dtypes
```

#### 필요한 날짜/시간 정보 추출

```python
# year, month, day, hour, minute, second

df['date_1'].dt.year
```

```python
## 요일 추출 -> (0: 월요일, 6: 일요일) weekday == dayofweek

df['date_1'].dt.weekday
```

#### 날짜 차이 계산

```python
(pd.to_datetime('today') - df['date_2']).dt.days
```

#### 날짜 관련 특징 생성

```python
## 단순 변환시 나노초 단위로 바뀌기 때문에 10^9을 나눠주어야 제대로 된 unix timestamp로 변환된다.

df['date_1'].astype(int)
```
