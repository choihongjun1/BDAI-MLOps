## 1. 텍스트 출력 위젯

````python
import streamlit as st

# 가장 크게 표시되는 페이지 제목
st.title('Streamlit 위젯 기초')

# 섹션 헤더 (제목보다 작음)
st.header('이것은 헤더입니다')

# 소제목 (헤더보다 작음)
st.subheader('이것은 서브헤더입니다')

# 작은 설명 텍스트 (회색)
st.caption('이것은 캡션입니다 — 보조 설명에 사용')

# 범용 출력 — 문자열, 마크다운, 숫자 등 뭐든 출력 가능
st.write('st.write는 **마크다운**도 됩니다')
st.write(42)          # 숫자도 출력
st.write([1, 2, 3])   # 리스트도 출력

# 마크다운 직접 작성
st.markdown('`st.markdown`으로 **굵게**, *기울여* 쓸 수 있어요')

# 코드 블록 표시
st.code('print("Hello World")', language='python')

# 수평 구분선
st.divider()
````

## 2. 텍스트 입력 위젯

````python
import streamlit as st

# 한 줄 텍스트 입력
name = st.text_input(
    label='이름을 입력하세요',        # 라벨
    value='',                         # 기본값
    placeholder='예: 홍길동',         # 힌트 텍스트
    max_chars=50,                      # 최대 입력 글자 수
)
st.write('입력한 이름:', name)

st.divider()

# 여러 줄 텍스트 입력
content = st.text_area(
    label='내용을 입력하세요',
    value='',
    height=150,                        # 입력창 높이 (픽셀)
    placeholder='자유롭게 입력...',
)
st.write('입력한 내용:', content)
````

## 3. 숫자 입력 위젯

````python
import streamlit as st

# 숫자 입력 박스 (위아래 화살표로 조절 가능)
age = st.number_input(
    label='나이를 입력하세요',
    min_value=0,    # 최솟값
    max_value=120,  # 최댓값
    value=25,       # 기본값
    step=1,         # 화살표 클릭 시 증감 단위
)
st.write('입력한 나이:', age)

st.divider()

# 슬라이더 (드래그로 값 선택)
score = st.slider(
    label='점수를 선택하세요',
    min_value=0,
    max_value=100,
    value=50,       # 기본값
    step=5,         # 이동 단위
)
st.write('선택한 점수:', score)

st.divider()

# 범위 슬라이더 (value를 튜플로 지정)
price_range = st.slider(
    label='가격 범위 (만원)',
    min_value=0,
    max_value=1000,
    value=(100, 500),   # (시작값, 끝값) 튜플
)
st.write('선택한 범위:', price_range)
````

## 4. 선택 위젯

````python
import streamlit as st

# 드롭다운 — 하나만 선택
fruit = st.selectbox(
    label='좋아하는 과일을 고르세요',
    options=['사과', '바나나', '딸기', '포도'],
    index=0,    # 기본 선택 인덱스
)
st.write('선택한 과일:', fruit)

st.divider()

# 멀티셀렉트 — 여러 개 선택 가능
skills = st.multiselect(
    label='보유 기술을 모두 선택하세요',
    options=['Python', 'SQL', 'Java', 'JavaScript', 'R'],
    default=['Python'],   # 기본 선택값 (리스트)
)
st.write('선택한 기술:', skills)

st.divider()

# 라디오 버튼 — 하나만 선택 (보기가 전부 노출)
gender = st.radio(
    label='성별을 선택하세요',
    options=['남성', '여성', '선택 안함'],
    index=0,
    horizontal=True,    # 가로로 배치
)
st.write('선택한 성별:', gender)

st.divider()

# 체크박스 — True / False 반환
agree = st.checkbox('이용약관에 동의합니다', value=False)
if agree:
    st.success('동의하셨습니다!')
````

## 5. 버튼 위젯

````python
import streamlit as st

# 기본 버튼 — 클릭하면 True 반환, 아니면 False
clicked = st.button('클릭해보세요')
if clicked:
    st.write('버튼이 클릭되었습니다!')

st.divider()

# type='primary' — 강조 색상 버튼
primary_btn = st.button('주요 액션', type='primary')

# use_container_width=True — 가로 꽉 채우기
full_btn = st.button('가로 꽉 채우기 버튼', use_container_width=True)

st.divider()

# 토글 버튼 — 눌릴 때마다 True/False 전환
toggled = st.toggle('다크모드', value=False)
if toggled:
    st.write('다크모드 ON')
else:
    st.write('다크모드 OFF')
````

## 6. 상태 메시지 위젯

````python
import streamlit as st

# 성공 (초록)
st.success('성공적으로 저장되었습니다!')

# 오류 (빨강)
st.error('오류가 발생했습니다. 다시 시도해주세요.')

# 경고 (노랑)
st.warning('입력값을 확인해주세요.')

# 정보 (파랑)
st.info('서버가 실행 중입니다.')

# 예외 출력 — 실제 파이썬 에러 객체를 보여줄 때 사용
try:
    1 / 0
except ZeroDivisionError as e:
    st.exception(e)
````

## 7. 레이아웃 위젯

````python
import streamlit as st

# ── columns: 화면을 여러 열로 분할 ──────────────────────────────────
col1, col2, col3 = st.columns(3)   # 동일 비율 3열

with col1:
    st.metric(label='정확도', value='92.5%', delta='+1.2%')  # delta: 변화량 표시
with col2:
    st.metric(label='손실', value='0.08', delta='-0.01')
with col3:
    st.metric(label='에포크', value='50')

st.divider()

# 비율 지정 열 분할 — [2, 1]: 왼쪽이 오른쪽의 2배 너비
left, right = st.columns([2, 1])
with left:
    st.write('넓은 열 (비율 2)')
with right:
    st.write('좁은 열 (비율 1)')

st.divider()

# ── container: 위젯을 논리적으로 묶는 블록 ──────────────────────────
with st.container():
    st.write('container 안의 내용')
    st.write('같은 container 안에 있어요')

st.divider()

# ── expander: 펼치기/접기 가능한 영역 ──────────────────────────────
with st.expander('자세히 보기 (클릭하면 펼쳐짐)', expanded=False):
    st.write('여기에 숨겨진 내용이 들어갑니다.')
    st.code('print("숨겨진 코드")')

st.divider()

# ── tabs: 탭 전환 레이아웃 ──────────────────────────────────────────
tab1, tab2 = st.tabs(['탭 A', '탭 B'])
with tab1:
    st.write('탭 A의 내용')
with tab2:
    st.write('탭 B의 내용')
````

## 8. 사이드바

````python
import streamlit as st

# ── 방법 1: with 구문 ──────────────────────────────────────────────
with st.sidebar:
    st.header('사이드바 설정')
    mode = st.selectbox('모드 선택', ['모드 A', '모드 B'])
    threshold = st.slider('임계값', 0.0, 1.0, 0.5)

# ── 방법 2: st.sidebar.위젯() 직접 호출 ───────────────────────────
show_detail = st.sidebar.checkbox('상세 정보 표시')

# 사이드바 값은 메인 화면에서 그냥 변수로 사용
st.write('선택한 모드:', mode)
st.write('설정한 임계값:', threshold)
````

## 9. 폼(Form) 위젯

````python
import streamlit as st

# Form: 내부 위젯 값을 한 번에 묶어서 제출
# 장점: submit 전까지 앱이 재실행되지 않아 불필요한 API 호출 방지

with st.form('my_form'):
    st.subheader('회원가입 폼')

    username = st.text_input('아이디')
    password = st.text_input('비밀번호', type='password')  # 입력값 마스킹
    age = st.number_input('나이', min_value=1, max_value=120, value=20)
    agree = st.checkbox('개인정보 수집에 동의합니다')

    # 폼 내부에는 반드시 form_submit_button 이 있어야 함
    submitted = st.form_submit_button('제출', type='primary', use_container_width=True)

# 제출 후 처리 — with 블록 바깥에서 작성
if submitted:
    if not username or not password:
        st.error('아이디와 비밀번호를 입력해주세요.')
    elif not agree:
        st.warning('개인정보 수집 동의가 필요합니다.')
    else:
        st.success(f'{username}님 가입 완료!')
````

## 10. session_state — 상태 유지

````python
import streamlit as st

# Streamlit은 위젯 조작 시마다 스크립트가 처음부터 재실행됩니다.
# session_state를 쓰면 재실행 사이에도 값을 기억할 수 있습니다.

# 초기값 설정 — 없을 때만 초기화 (중요!)
if 'count' not in st.session_state:
    st.session_state['count'] = 0

if 'history' not in st.session_state:
    st.session_state['history'] = []

# 버튼 클릭 시 count 증가
if st.button('카운트 +1'):
    st.session_state['count'] += 1
    st.session_state['history'].append(st.session_state['count'])

if st.button('초기화'):
    st.session_state['count'] = 0
    st.session_state['history'] = []

st.metric('현재 카운트', st.session_state['count'])
st.write('기록:', st.session_state['history'])
````

## 11. 스피너 & 프로그레스바

````python
import streamlit as st
import time

# spinner: 작업 중임을 알려주는 로딩 표시
if st.button('작업 시작'):
    with st.spinner('처리 중입니다...'):
        time.sleep(2)   # 실제 코드에선 API 호출, 모델 예측 등
    st.success('완료!')

st.divider()

# progress bar: 진행률 표시 (0.0 ~ 1.0)
if st.button('진행률 시뮬레이션'):
    bar = st.progress(0, text='시작 중...')
    for i in range(1, 11):
        time.sleep(0.1)
        bar.progress(i / 10, text=f'진행 중... {i * 10}%')
    st.success('완료!')
````

## 12. 파일 업로드 & 날짜/시간 선택

````python
import streamlit as st
import pandas as pd

# ── 파일 업로드 ────────────────────────────────────────────────────
uploaded = st.file_uploader(
    label='CSV 파일을 업로드하세요',
    type=['csv'],             # 허용 확장자 리스트
    accept_multiple_files=False,
)
if uploaded:
    df = pd.read_csv(uploaded)
    st.write('파일 미리보기:', df.head())

st.divider()

# ── 날짜 선택 ──────────────────────────────────────────────────────
from datetime import date

selected_date = st.date_input(
    label='날짜를 선택하세요',
    value=date.today(),
)
st.write('선택한 날짜:', selected_date)

st.divider()

# ── 시간 선택 ──────────────────────────────────────────────────────
from datetime import time

selected_time = st.time_input(
    label='시간을 선택하세요',
    value=time(9, 0),   # 기본값: 09:00
)
st.write('선택한 시간:', selected_time)
````
